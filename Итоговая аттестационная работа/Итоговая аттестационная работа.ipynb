### Функция `is_valid_class_folder`

Функция предназначена для фильтрации недопустимых (служебных или скрытых) директорий при анализе структуры датасета.  
Она исключает папки, которые:

- являются системными или служебными (`__pycache__`, `.git`, `.ipynb_checkpoints` и др.);
- начинаются с точки (скрытые файлы/каталоги в Unix-системах);
- не относятся к классам данных.

Таким образом, функция помогает корректно определить папки, соответствующие настоящим классам изображений.

````python
def is_valid_class_folder(folder_name):
    """Проверка, является ли папка валидным классом"""
    invalid_folders = ['.ipynb_checkpoints', '__pycache__', '.git', 'Thumbs.db', '.DS_Store']
    return folder_name not in invalid_folders and not folder_name.startswith('.')
````

### Настройка генератора обучающих данных

Создаётся объект `ImageDataGenerator` из библиотеки Keras/TensorFlow для аугментации и предобработки изображений:

- **`rescale=1./255`** — нормализация пикселей из диапазона [0, 255] в [0, 1];  
- **`rotation_range=20`** — случайный поворот изображения до ±20 градусов;  
- **`width_shift_range=0.2`**, **`height_shift_range=0.2`** — случайный сдвиг по горизонтали и вертикали до 20% ширины/высоты;  
- **`shear_range=0.2`** — случайное сдвиговое («перекосное») преобразование;  
- **`zoom_range=0.2`** — случайное масштабирование (приближение/отдаление) до ±20%;  
- **`horizontal_flip=True`** — разрешено случайное горизонтальное отражение (зеркальный поворот);  
- **`fill_mode='nearest'`** — способ заполнения пустых пикселей после трансформаций (используется ближайший сосед);  
- **`validation_split=0.2`** — 20% обучающих данных автоматически выделяются для валидации.

Этот генератор будет использоваться для создания обучающей и валидационной выборок из папки `train`.

````python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=0.2  
)
````

### Создание простой свёрточной нейронной сети (CNN)

Функция `create_simple_cnn` определяет базовую архитектуру CNN для задачи классификации изображений:

- **Вход**: изображения заданного размера (`input_shape`);
- **Свёрточные блоки**: три последовательных блока, каждый состоит из:
  - свёрточного слоя (`Conv2D`) с увеличением числа фильтров (32 → 64 → 128);
  - функции активации ReLU;
  - подвыборки (`MaxPooling2D`) с уменьшением пространственного размера в 2 раза;
- **Полносвязная часть**:
  - выравнивание признаков (`Flatten`);
  - скрытый полносвязный слой на 128 нейронов с ReLU;
  - регуляризация через `Dropout(0.5)` для борьбы с переобучением;
- **Выходной слой**: `num_classes` нейронов с активацией `softmax` для многоклассовой классификации.

Эта архитектура сбалансирована для небольших датасетов и служит хорошей отправной точкой для экспериментов.

````python
def create_simple_cnn(input_shape, num_classes):
    model = models.Sequential([
        layers.Conv2D(32, (3, 3), activation='relu', input_shape=input_shape),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(64, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Conv2D(128, (3, 3), activation='relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model
````

### Создание углублённой свёрточной нейронной сети (Deep CNN)

Функция `create_deep_cnn` реализует более мощную и регуляризованную архитектуру CNN, подходящую для задач с умеренным объёмом данных:

- **Блоки свёрток**: три этапа с глубиной 32 → 64 → 128 фильтров;  
  каждый этап содержит **две свёртки подряд** с `padding='same'`, чтобы сохранять пространственное разрешение;
- **Нормализация и активация**: после каждого свёрточного слоя применяется `BatchNormalization` и `ReLU` — это ускоряет обучение и повышает стабильность;
- **Регуляризация**:  
  - `Dropout(0.25)` после каждого блока понижения размерности;  
  - `Dropout(0.5)` перед выходным слоем;
- **Полносвязная часть**:  
  - скрытый слой на 256 нейронов с ReLU;  
  - дополнительная `BatchNormalization` для улучшения обобщения;
- **Выход**: слой с `num_classes` нейронами и активацией `softmax` для многоклассовой классификации.

Эта модель лучше справляется с переобучением и может извлекать более сложные признаки по сравнению с простой CNN.

````python
def create_deep_cnn(input_shape, num_classes):
    model = models.Sequential([
        layers.Conv2D(32, (3, 3), padding='same', input_shape=input_shape),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Conv2D(32, (3, 3), padding='same'),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        layers.Conv2D(64, (3, 3), padding='same'),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Conv2D(64, (3, 3), padding='same'),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        layers.Conv2D(128, (3, 3), padding='same'),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Conv2D(128, (3, 3), padding='same'),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),
        
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model
````

### Создание модели на основе трансферного обучения (MobileNetV2)

Функция `create_transfer_learning_model` строит классификатор, используя предобученную сеть **MobileNetV2** в качестве основы:

- **Базовая модель**:  
  - загружается с весами, предобученными на ImageNet (`weights='imagenet'`);  
  - финальный полносвязный слой (`include_top=False`) отключён, чтобы заменить его собственной головой классификации.
- **Тонкая настройка (fine-tuning)**:  
  - сначала делается всё дерево модели **обучаемым** (`base_model.trainable = True`);  
  - затем замораживаются все слои, кроме последних `trainable_layers` (по умолчанию — 20), чтобы сохранить общие признаки и адаптировать только верхние уровни к целевой задаче.
- **Голова классификатора**:  
  - `GlobalAveragePooling2D` уменьшает пространственные карты признаков до одного вектора на изображение;  
  - полносвязный слой с 256 нейронами и ReLU;  
  - регуляризация через `Dropout(0.5)`;  
  - выходной слой с активацией `softmax` для `num_classes` классов.

Такой подход позволяет эффективно обучать высокоточные модели даже при ограниченном объёме данных.

````python
def create_transfer_learning_model(input_shape, num_classes, trainable_layers=20):
    base_model = MobileNetV2(
        input_shape=input_shape,
        include_top=False,
        weights='imagenet'
    )
    

    base_model.trainable = True
    for layer in base_model.layers[:-trainable_layers]:
        layer.trainable = False
    
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(num_classes, activation='softmax')
    ])
    return model
````

### Создание продвинутой свёрточной нейронной сети (Advanced CNN)

Функция `create_advanced_cnn` реализует усовершенствованную архитектуру CNN с использованием функционального API Keras:

- **Входной слой**: принимает изображения заданного `input_shape`.
- **Три последовательных блока признаков** с нарастающей глубиной фильтров (64 → 128 → 256):  
  - каждый блок содержит две свёртки с `padding='same'`, чтобы сохранять пространственное разрешение;  
  - после каждой свёртки применяются `BatchNormalization` и `ReLU`;  
  - завершается `MaxPooling2D` для понижения размерности и `Dropout` с прогрессивно увеличивающимся коэффициентом (0.3 → 0.4) для регуляризации.
- **Голова классификатора**:  
  - используется `GlobalAveragePooling2D` вместо `Flatten` для снижения числа параметров и повышения обобщающей способности;  
  - два полносвязных слоя (512 → 256 нейронов) с ReLU, `BatchNormalization` и `Dropout(0.5)` между ними;  
  - финальный слой с `softmax`-активацией для многоклассовой классификации.
- **Итоговая модель**: строится как объект `keras.Model` с явным указанием входа и выхода.

Эта архитектура обеспечивает баланс между выразительной мощностью и устойчивостью к переобучению, особенно на средних по размеру датасетах.

````python
def create_advanced_cnn(input_shape, num_classes):
    inputs = keras.Input(shape=input_shape)
    

    x = layers.Conv2D(64, (3, 3), padding='same')(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(64, (3, 3), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)
    

    x = layers.Conv2D(128, (3, 3), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(128, (3, 3), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)
    

    x = layers.Conv2D(256, (3, 3), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(256, (3, 3), padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.4)(x)
    

    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(512, activation='relu')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(0.5)(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)
    
    model = keras.Model(inputs=inputs, outputs=outputs)
    return model
````

### Обучение и оценка модели

Функция `train_and_evaluate` реализует полный цикл обучения, валидации и тестирования нейросетевой модели:

- **Настройка оптимизатора**: поддерживается выбор между `Adam`, `SGD` (с импульсом) и `RMSprop`; скорость обучения (`learning_rate`) задаётся параметром.
- **Компиляция модели**: используется кросс-энтропийная функция потерь (`categorical_crossentropy`) и метрика точности (`accuracy`).
- **Коллбэки для устойчивого обучения**:
  - `EarlyStopping`: останавливает обучение, если `val_loss` не улучшается в течение 7 эпох, и восстанавливает лучшие веса;
  - `ReduceLROnPlateau`: уменьшает скорость обучения в 2 раза (`factor=0.5`), если `val_loss` не улучшается 3 эпохи подряд, но не ниже `1e-7`.
- **Обучение**: выполняется на `train_generator` с валидацией на `validation_generator` в течение `EPOCHS` эпох.
- **Тестирование**: после обучения модель оценивается на `test_generator`, выводятся итоговые значения потерь и точности.
- **Визуализация**:
  - `plot_training_history`: строит графики динамики точности и потерь на обучающей и валидационной выборках;
  - `plot_confusion_matrix`: отображает матрицу ошибок для анализа качества классификации по каждому классу.

Функция возвращает историю обучения и итоговую точность на тестовой выборке, что удобно для сравнения разных архитектур или гиперпараметров.

````python
def train_and_evaluate(model, model_name, optimizer='adam', learning_rate=0.001):
    print(f"\n{'='*60}")
    print(f"Обучение модели: {model_name}")
    print(f"{'='*60}\n")
    
    if optimizer == 'adam':
        opt = keras.optimizers.Adam(learning_rate=learning_rate)
    elif optimizer == 'sgd':
        opt = keras.optimizers.SGD(learning_rate=learning_rate, momentum=0.9)
    elif optimizer == 'rmsprop':
        opt = keras.optimizers.RMSprop(learning_rate=learning_rate)
    
    model.compile(
        optimizer=opt,
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    
    early_stop = EarlyStopping(
        monitor='val_loss',
        patience=7,  
        restore_best_weights=True,
        verbose=1
    )
    
    reduce_lr = ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3, 
        min_lr=1e-7,
        verbose=1
    )
    

    history = model.fit(
        train_generator,
        epochs=EPOCHS,
        validation_data=validation_generator,
        callbacks=[early_stop, reduce_lr],
        verbose=1
    )
    

    print("\nОценка на тестовой выборке:")
    test_loss, test_accuracy = model.evaluate(test_generator)
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {test_accuracy:.4f}")
    

    plot_training_history(history, model_name)
    

    plot_confusion_matrix(model, test_generator, class_names, model_name)
    
    return history, test_accuracy
````

### Визуализация истории обучения

Функция `plot_training_history` отображает графики изменения **точности** и **функции потерь** в процессе обучения модели:

- Создаётся фигура с двумя подграфиками:  
  - **Слева**: динамика точности (`accuracy`) на обучающей и валидационной выборках;  
  - **Справа**: динамика потерь (`loss`) на тех же выборках.
- По каждой оси добавляются подписи, заголовок с названием модели, легенда и сетка для удобства анализа.
- Графики позволяют быстро оценить:
  - сходится ли обучение;
  - есть ли признаки переобучения (например, когда точность на трейне растёт, а на валидации — падает);
  - насколько стабильно изменяется ошибка.

Результат выводится с помощью `plt.show()`, обеспечивая наглядную обратную связь при экспериментировании с моделями.

````python
def plot_training_history(history, model_name):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    

    ax1.plot(history.history['accuracy'], label='Train Accuracy')
    ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
    ax1.set_title(f'{model_name} - Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()
    ax1.grid(True)
    

    ax2.plot(history.history['loss'], label='Train Loss')
    ax2.plot(history.history['val_loss'], label='Val Loss')
    ax2.set_title(f'{model_name} - Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True)
    
    plt.tight_layout()
    plt.show()
````

### Построение матрицы ошибок и отчёт классификации

Функция `plot_confusion_matrix` выполняет детальную оценку качества модели на тестовой выборке:

- **Сброс генератора**: `test_gen.reset()` гарантирует, что все изображения будут обработаны с самого начала и в правильном порядке.
- **Получение предсказаний**: модель делает прогнозы для всех батчей, после чего извлекаются индексы наиболее вероятных классов (`y_pred`).
- **Истинные метки**: берутся напрямую из атрибута `test_gen.classes`.
- **Матрица ошибок**: строится с помощью `confusion_matrix` из scikit-learn и визуализируется тепловой картой (`sns.heatmap`) с подписями классов по осям.
- **Отчёт классификации**: выводится подробный текстовый отчёт с метриками precision, recall, f1-score и support для каждого класса с помощью `classification_report`.

Этот анализ помогает выявить, какие классы модель путает чаще всего, и оценить её сбалансированность по всем категориям.

````python
def plot_confusion_matrix(model, test_gen, class_names, model_name):
    test_gen.reset()
    predictions = model.predict(test_gen, verbose=1)
    y_pred = np.argmax(predictions, axis=1)
    y_true = test_gen.classes
    
    cm = confusion_matrix(y_true, y_pred)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names,
                yticklabels=class_names)
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.show()
    
    print(f"\nClassification Report для {model_name}:")
    print(classification_report(y_true, y_pred, target_names=class_names))
````

### Последовательный запуск экспериментов с разными архитектурами

Этот блок кода реализует управляемый пайплайн для сравнительного анализа нескольких моделей:

- **Инициализация**:  
  - определяется форма входного тензора (`input_shape = (128, 128, 3)`);  
  - создаётся словарь `results` для хранения итоговой точности каждой модели;  
  - выводится информация о числе и названиях классов, а также о размере изображений.

- **Условное выполнение**:  
  Каждая модель запускается только если соответствующий флаг (`RUN_...`) установлен в `True`.  
  Для каждой активной модели:
  - выводится заголовок эксперимента;
  - создаётся архитектура вызовом соответствующей функции;
  - отображается краткая сводка модели (`model.summary()`);
  - выполняется обучение и оценка через `train_and_evaluate`;
  - точность на тестовой выборке сохраняется в `results`.

- **Особенности экспериментов**:
  - **Simple CNN**: простейшая архитектура для базового сравнения;
  - **Deep CNN**: усиленная регуляризация и нормализация;
  - **Transfer Learning**: использует MobileNetV2 с пониженной скоростью обучения (`1e-4`);
  - **Advanced CNN**: более глубокая и регуляризованная сеть на функциональном API;
  - **Deep CNN + SGD**: та же архитектура, но с оптимизатором SGD и повышенной скоростью обучения (`1e-2`) для изучения влияния алгоритма оптимизации.

Такой подход позволяет систематически сравнить эффективность различных стратегий (простая CNN vs. transfer learning, Adam vs. SGD и т.д.) в одинаковых условиях.

````python
input_shape = (*IMG_SIZE, 3)
results = {}

print(f"\nИспользуется {NUM_CLASSES} классов: {class_names}")
print(f"Входной размер изображения: {input_shape}\n")


if RUN_SIMPLE_CNN:
    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ 1: Простая CNN")
    print("="*60)
    model1 = create_simple_cnn(input_shape, NUM_CLASSES)
    model1.summary()
    history1, acc1 = train_and_evaluate(model1, "Simple CNN", optimizer='adam', learning_rate=0.001)
    results['Simple CNN'] = acc1


if RUN_DEEP_CNN:
    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ 2: Глубокая CNN с BatchNormalization")
    print("="*60)
    model2 = create_deep_cnn(input_shape, NUM_CLASSES)
    model2.summary()
    history2, acc2 = train_and_evaluate(model2, "Deep CNN", optimizer='adam', learning_rate=0.001)
    results['Deep CNN'] = acc2


if RUN_TRANSFER_LEARNING:
    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ 3: Transfer Learning (MobileNetV2)")
    print("="*60)
    model3 = create_transfer_learning_model(input_shape, NUM_CLASSES)
    model3.summary()
    history3, acc3 = train_and_evaluate(model3, "Transfer Learning", optimizer='adam', learning_rate=0.0001)
    results['Transfer Learning'] = acc3


if RUN_ADVANCED_CNN:
    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ 4: Улучшенная CNN")
    print("="*60)
    model4 = create_advanced_cnn(input_shape, NUM_CLASSES)
    model4.summary()
    history4, acc4 = train_and_evaluate(model4, "Advanced CNN", optimizer='adam', learning_rate=0.001)
    results['Advanced CNN'] = acc4


if RUN_DEEP_CNN_SGD:
    print("\n" + "="*60)
    print("ЭКСПЕРИМЕНТ 5: Deep CNN с SGD оптимизатором")
    print("="*60)
    model5 = create_deep_cnn(input_shape, NUM_CLASSES)
    history5, acc5 = train_and_evaluate(model5, "Deep CNN + SGD", optimizer='sgd', learning_rate=0.01)
    results['Deep CNN + SGD'] = acc5
````
